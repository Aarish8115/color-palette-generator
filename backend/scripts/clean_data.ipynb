```python
def get_hue_family(h, s, l):
    if s < 10:
        return "neutral"
    if 15 <= h <= 45 and s <= 60 and l <= 55:
        return "brown"
    if h < 15 or h >= 345:
        return "red"
    elif h < 45:
        return "orange"
    elif h < 70:
        return "yellow"
    elif h < 170:
        return "green"
    elif h < 200:
        return "cyan"
    elif h < 260:
        return "blue"
    elif h < 300:
        return "purple"
    elif h < 345:
        return "pink"

data["hue_family"] = data.apply(
    lambda row: get_hue_family(row["Hue"], row["Saturation"], row["Lightness"]),
    axis=1
)
```

```python
import re

def normalize_text(text):
    if pd.isna(text):
        return text
    return re.sub(r"[^\w\s]", "", text.lower())
```

```python
exclude = [
    "HEX Code", "Keywords", "R", "G", "B",
    "Hue", "Saturation", "Lightness"
]

columns = [col for col in data.columns if col not in exclude]

data[columns] = data[columns].apply(lambda col: col.map(normalize_text))
```

```python
data["Keywords"] = data["Keywords"].apply(
    lambda x: [word.strip().lower() for word in x.split(",")] 
    if isinstance(x, str) else []
)
```

```python
data=data.rename(columns={
    "Color Name":'color_name',
    'HEX Code':'hex',
    'Category':'category',
    'Keywords':'keywords'
})
```

```python
import matplotlib.colors as mcolors

data["rgb"] = data["hex"].apply(
    lambda x: tuple(int(c * 255) for c in mcolors.to_rgb(x))
)
```

```python
import colorsys

def hex_to_hsl(hex_code):
    hex_code = hex_code.lstrip("#")
    r, g, b = [int(hex_code[i:i+2], 16) / 255 for i in (0, 2, 4)]

    h, l, s = colorsys.rgb_to_hls(r, g, b)

    return (round(h * 360), round(s * 100), round(l * 100))

data["hsl"] = data["hex"].apply(hex_to_hsl)
```

```python
def get_brightness(r, g, b):
    # Normalize RGB
    r, g, b = r / 255, g / 255, b / 255

    # sRGB → linear RGB
    def linear(c):
        return c / 12.92 if c <= 0.04045 else ((c + 0.055) / 1.055) ** 2.4

    r, g, b = map(linear, (r, g, b))

    # Relative luminance
    luminance = 0.2126 * r + 0.7152 * g + 0.0722 * b

    if luminance >= 0.7:
        return "bright"
    elif luminance <= 0.25:
        return "dark"
    else:
        return "medium"
```

```python
data["brightness"] = data["rgb"].apply(
    lambda rgb: get_brightness(*rgb)
)
```

```python
def get_warmth(r, g, b):
    warm_score = r - b

    if warm_score > 40:
        return "warm"
    elif warm_score < -40:
        return "cool"
    else:
        return "neutral"
```

```python
data["warmth"] = data["rgb"].apply(
    lambda rgb: get_warmth(*rgb)
)
```

```python
def contrast_score(r, g, b):
    r, g, b = r / 255, g / 255, b / 255

    def linear(c):
        return c / 12.92 if c <= 0.04045 else ((c + 0.055) / 1.055) ** 2.4

    r, g, b = map(linear, (r, g, b))

    luminance = 0.2126 * r + 0.7152 * g + 0.0722 * b

    contrast_white = 1.05 / (luminance + 0.05)
    contrast_black = (luminance + 0.05) / 0.05

    return round(max(contrast_white, contrast_black), 2)
```

```python
data["contrast_score"] = data["rgb"].apply(
    lambda rgb: contrast_score(*rgb)
)
```

```python
def get_color_style(h, s, l):
    styles = []

    # Pastel
    if s >= 20 and s <= 60 and l >= 70:
        styles.append("pastel")

    # Vibrant
    if s >= 70 and l >= 40 and l <= 65:
        styles.append("vibrant")

    # Muted
    if s < 40 and l >= 30 and l <= 70:
        styles.append("muted")

    # Deep
    if l < 30 and s >= 40:
        styles.append("deep")

    # Bright
    if l >= 75 and s >= 40:
        styles.append("bright")

    # Soft
    if s < 50 and l >= 60:
        styles.append("soft")

    # Bold
    if s >= 75:
        styles.append("bold")

    return styles
```

```python
data["color_style"] = data['hsl'].apply(
    lambda hsl: get_color_style(
        *hsl
    ),
)
```
